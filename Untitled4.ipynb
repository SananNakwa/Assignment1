```python
from transformers import AutoModelForCausalLM, AutoTokenizer
from datasets import load_dataset
from trl import SFTConfig, SFTTrainer, setup_chat_format
import torch

device = (
    "cuda"
    if torch.cuda.is_available()
    else "mps" if torch.backends.mps.is_available() else "cpu"
)

# Load the model and tokenizer
model_name = "HuggingFaceTB/SmolLM2-135M"
model = AutoModelForCausalLM.from_pretrained(
    pretrained_model_name_or_path=model_name
).to(device)
tokenizer = AutoTokenizer.from_pretrained(pretrained_model_name_or_path=model_name)

# Set up the chat format
model, tokenizer = setup_chat_format(model=model, tokenizer=tokenizer)

# Set our name for the finetune to be saved &/ uploaded to
finetune_name = "SmolLM2-FT-MyDataset"
finetune_tags = ["smol-course", "module_1"]
```

```python
prompt = "what is optical fiber"

# Format with template
messages = [{"role": "user", "content": prompt}]
formatted_prompt = tokenizer.apply_chat_template(messages, tokenize=False)

# Generate response
inputs = tokenizer(formatted_prompt, return_tensors="pt").to(device)
outputs = model.generate(**inputs, max_new_tokens=100)
print("Before training:")
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
# Load a sample dataset
from datasets import load_dataset

# TODO: define your dataset and config using the path and name parameters
ds = load_dataset(path="Treza12/Dataset_Optical_Fiber2")
```

```python
# Create a 'text' field by concatenating 'Question' and 'Answer'
def add_text_field(example):
    # Ensure that both 'Question' and 'Answer' are strings
    question = str(example['Question'])
    answer = str(example['Answer'])

    # Concatenate the two fields
    example['text'] = question + " " + answer  # Modify the format as needed

    return example

# Apply the function to the dataset
ds = ds.map(add_text_field)

# Now your dataset will have a 'text' field
```

```python
# Configure the SFTConfig for training on CPU
sft_config = SFTConfig(
    output_dir="./sft_output",
    max_steps=1000,  # Adjust based on dataset size and desired training duration
    per_device_train_batch_size=4,  # Adjust batch size based on your CPU's capacity
    learning_rate=5e-5,  # Common starting point for fine-tuning
    logging_steps=10,  # Frequency of logging training metrics
    save_steps=100,  # Frequency of saving model checkpoints
    evaluation_strategy="steps",  # Evaluate the model at regular intervals
    eval_steps=50,  # Frequency of evaluation
    use_mps_device=(
        True if device == "mps" else False
    ),
    hub_model_id=finetune_name,  # Set a unique name for your model
)

# Create subsets of the train and test datasets
train_subset = ds["train"].select(range(2000))  # First 2000 samples for training
test_subset = ds["test"].select(range(1000))  # First 1000 samples for testing

# Initialize the SFTTrainer
trainer = SFTTrainer(
    model=model,
    args=sft_config,
    train_dataset=train_subset,
    tokenizer=tokenizer,
    eval_dataset=test_subset,
)
```

```python
# Train the model
trainer.train()

# Save the model
trainer.save_model(f"./{finetune_name}")
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

# Load the fine-tuned model and tokenizer
model = AutoModelForCausalLM.from_pretrained("SmolLM2-FT-MyDataset")  # path to the fine-tuned model
tokenizer = AutoTokenizer.from_pretrained("SmolLM2-FT-MyDataset")

# Function to generate a response based on a prompt
def generate_response(prompt: str, max_length: int = 100):
    # Encode the prompt text to input IDs
    input_ids = tokenizer.encode(prompt, return_tensors="pt")

    # Generate response with sampling enabled (for randomness)
    output = model.generate(input_ids,
                            max_length=max_length,
                            num_return_sequences=1,
                            do_sample=True,  # Enable sampling for diverse responses
                            no_repeat_ngram_size=2,
                            top_p=0.95,
                            temperature=0.7)

    # Decode the generated response back to text
    response = tokenizer.decode(output[0], skip_special_tokens=True)

    return response

# Example usage
prompt = "What is the optical fiber communication technology?"
response = generate_response(prompt)
print(response)
```
